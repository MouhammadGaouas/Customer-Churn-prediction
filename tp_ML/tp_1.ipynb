```python
import numpy as np 

X = np.array([
    [1, 0],
    [1, 0.1],
    [1, 0.2],
    [1, 0.3],
    [1, 0.4]
])
```

```python
F = np.array([0, 10, 20, 30, 40])

coefficients = np.linalg.inv(X.T @ X) @ X.T @ F

print(f"Coefficients : {coefficients}")
```

```python
F_pred = X @ coefficients
print(f"Predictions : {F_pred}")
print(f"Vraies valeurs : {F}")
print(f"Erreur (F - F_pred) : {F - F_pred}")
```
